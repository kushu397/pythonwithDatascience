### Create JOSN file

```python

#open(filename,operation)
import json
products_json = [
  {
    "Product Name": "Laptop",
    "Product Price": "$10000",
    "Product Brand": "HP"
  },
  {
    "Product Name": "Laptop",
    "Product Price": "$30000",
    "Product Brand": "Lenovo"
  },
  {
    "Product Name": "Laptop",
    "Product Price": "$20000",
    "Product Brand": "Dell"
  }
]

with open('products.json','w') as file:
    json.dump(products_json,file,indent=4)

```

### Read Json file

```python


import json
products_json = [
  {
    "Product Name": "Laptop",
    "Product Price": "$10000",
    "Product Brand": "HP"
  },
  {
    "Product Name": "Laptop",
    "Product Price": "$30000",
    "Product Brand": "Lenovo"
  },
  {
    "Product Name": "Laptop",
    "Product Price": "$20000",
    "Product Brand": "Dell"
  }
]

with open('products.json','r') as file:
    data = json.load(file)
    for row in data:
        for key , val in row.items():
            print(f'{key}:{val}')
        print('\n')
```

```python
from PyPDF2 import PdfReader, PdfWriter

reader = PdfReader("Kushal Adhikari.pdf")
writer = PdfWriter()

# Copy all pages to the writer object
for page in reader.pages:
    writer.add_page(page)

# Save to a NEW file name to prevent corrupting the source
with open("Kushal_Adhikari_Copy.pdf", "wb") as output_file:
    writer.write(output_file)

print("PDF file copied to Kushal_Adhikari_Copy.pdf successfully!")
```
