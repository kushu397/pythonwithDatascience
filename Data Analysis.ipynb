```python
import pandas as pd


def currencyConverter(dataframe: pd.DataFrame, prefix: str, *args: str):
  for col in args:
    # Use a lambda function to properly format each value in the column
    dataframe[col] = dataframe[col].map(lambda x: f'{prefix}{x:,.2f}')
        
```

```python
emp_details = (
    df.groupby('Department')
    .agg(
        total_employees=('EmpID', 'count'),
        total_salary=('Salary', 'sum'),
        Average_salary=('Salary', 'mean'),
    )
    .reset_index()
)

emp_details.columns = emp_details.columns.str.replace('_', ' ', regex=False)

# Pass the exact column names generated ('total salary', 'Average salary')
currencyConverter(emp_details, '$', 'total salary', 'Average salary')

emp_details
```

```python
### Find total number of employee hired in each year.
import pandas as pd

# Extract the year from 'DateofHire' to create 'Hire_Year' dynamically
df['Hire_Year'] = pd.to_datetime(df['DateofHire']).dt.year

# --- Your Code ---
hire_year_emp = (
    df.groupby('Hire_Year')['EmpID']
    .count()
    .reset_index()
    .rename(
        columns={'Hire_Year': 'Hire year', 'EmpID': 'Total Employees'}
    )
)

hire_year_emp['Hire%'] = (
    hire_year_emp['Total Employees'].pct_change().fillna(0)
) * 100
hire_year_emp['Hire%'] = hire_year_emp['Hire%'].map('{:,.2f}%'.format)

hire_year_emp
```

```python
## Find total number of employees hired in each month
import pandas as pd

# 1. Extract month number and month name from 'DateofHire' first
df['Hire_Month'] = pd.to_datetime(df['DateofHire']).dt.month
df['Hire_Month_Name'] = pd.to_datetime(df['DateofHire']).dt.strftime('%B')

# 2. Your Code (fixed 'Toral Employees' typo to match your column selection)
emp_hire_month = (
    df.groupby(['Hire_Month', 'Hire_Month_Name'])['EmpID']
    .count()
    .reset_index()
    .rename(
        columns={
            'Hire_Month': 'Hire Month',
            'EmpID': 'Total Employees',  # Fixed typo from 'Toral'
            'Hire_Month_Name': 'Month Name',
        }
    )
)[['Month Name', 'Total Employees']]

emp_hire_month
```

```python
import pandas as pd

# 1. Extract the day name from DateofHire first
df['Hire_Day_Name'] = pd.to_datetime(df['DateofHire']).dt.day_name()

# 2. Your Code (fixed pd.categorical to pd.Categorical)
weekdays = [
    'Sunday',
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
]
df['Hire_Day_Name'] = pd.Categorical(df['Hire_Day_Name'], categories=weekdays)

emp_hire_day = (
    df.groupby(['Hire_Day_Name'], observed=False)['EmpID'].count().reset_index()
)
emp_hire_day
```
