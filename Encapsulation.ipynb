# private access modifier

```python
class Category:
    def __init__(self, category_id, category_name):
        self.__category_id = category_id
        self.__category_name = category_name

    def orderCategories(self):
        print(f"Category_id: {self.__category_id}")
        print(f"Category_name: {self.__category_name}")


class Brand:
    def __init__(self, brand_id, brand_name):
        self.__brand_id = brand_id
        self.__brand_name = brand_name

    def brandDetails(self):
        print(f"Brand_id: {self.__brand_id}")
        print(f"Brand_name: {self.__brand_name}")


class Product(Category, Brand):

    def __init__(
        self,
        product_id,
        product_name,
        product_desc,
        product_price,
        category_id,
        category_name,
        brand_id,
        brand_name
    ):

        # Calling parent class constructors
        Category.__init__(self, category_id, category_name)
        Brand.__init__(self, brand_id, brand_name)

        # Private variables
        self.__product_id = product_id
        self.__product_name = product_name
        self.__product_desc = product_desc
        self.__product_price = product_price

    def product_details(self):
        print(f"Product_id: {self.__product_id}")
        print(f"Product_name: {self.__product_name}")
        print(f"Product_desc: {self.__product_desc}")
        print(f"Product_price: {self.__product_price}")


# Creating Product object
p = Product(
    'p-101',
    'Mobile',
    'Smart Mobile Phone',
    89000,
    2,
    'Electronics',
    'b-101',
    'Anker'
)

# Calling methods
p.product_details()
p.brandDetails()
p.orderCategories()
```

```python
"""
Access Modifiers/control:
Decides which data can be accessed from outside the class and which data should be hidden

public: accessible everywhere (default)
protected: accessible in class and child class (syntax: _variable)
private: accessible only inside same class (syntax: __variable, __method_name() )
"""
```

```python


class Demo:
    def __init__(self):
        self.name = 'Ram'         # Public
        self._age = 20            # Protected (convention)
        self.__salary = 50000     # Private (Name mangled)

    def show_name(self):
        print(self.name)

    def _show_age(self):          # Single underscore matching call
        print(self._age)

    def __show_salary(self):
        print(self.__salary)

    def show(self):
        self.__show_salary()      # Accesses private method inside class

d = Demo()

# Direct attribute access
print(d.name)       # Output: Ram
print(d._age)       # Output: 20 (Works, but convention discourages direct outside access)
# print(d.__salary) # Error: AttributeError (Private)

# Method calls
d.show_name()       # Output: Ram
d._show_age()       # Output: 20
d.show()            # Output: 50000 (Calls private method from inside class)
```
