### Control statement

```python
#odd and even
'''
annual salary -< = 200000=> 0% tax
                  250001 - 500000->10%
                  500001 - 1000000->15%
                  10000001>->30%


                  annaul salary
                  taxable income
                  remaining income
                  
                        

''' 
```

```python
tax = 466000
tax_amount = 0

if tax <= 200000:
    tax_amount = 0

elif tax <= 500000:
    tax_amount = (tax * 10) / 100

elif tax <= 1000000:
    tax_amount = (tax * 15) / 100

else:
    tax_amount = (tax * 30) / 100

remaining_income = tax - tax_amount

print(f"Your annual salary is = {tax}")
print(f"Your tax amount is = {tax_amount}") 
print(f"Your remaining income is = {remaining_income}")
```

```python
annual_income = int(input('Enter your annual income: '))
if annual_income >= 0 and annual_income <= 250000:

    print('0% - No Tax')
    taxable_income = 0
    remaining_balance = annual_income

    print(f'Annual Income: {annual_income}')
    print(f'Taxable Income: {taxable_income}')
    print(f'Remaining Income: {remaining_balance}')


elif annual_income > 250000 and annual_income <= 500000:
    print('10% Tax')
    taxable_income = annual_income * 0.10
    remaining_balance = annual_income - taxable_income

    print(f'Annual Income: {annual_income}')
    print(f'Taxable Income: {taxable_income}')
    print(f'Remaining Income: {remaining_balance}')


elif annual_income > 500000 and annual_income <= 1000000:
    print('15% Tax')
    taxable_income = annual_income * 0.15
    remaining_balance = annual_income - taxable_income

    print(f'Annual Income: {annual_income}')
    print(f'Taxable Income: {taxable_income}')
    print(f'Remaining Income: {remaining_balance}')


elif annual_income > 1000000:
    print('30% Tax')
    taxable_income = annual_income * 0.30
    remaining_balance = annual_income - taxable_income

    print(f'Annual Income: {annual_income}')
    print(f'Taxable Income: {taxable_income}')
    print(f'Remaining Income: {remaining_balance}')


else:
    print('Invalid income amount.')
```

### Nested if

```python
username = 'abc@gmail.com'
password = '12345678'

uname = input('Enter your username: ')
pwd = input('Enter your password: ')

if uname != username and pwd != password:
    print('Username or password incorrect.')

else:
    if uname == username:
        if pwd == password:
            print('Login successful.')
        else:
            print('Incorrect password.')
    else:
        print('User not found with given username.')
```
