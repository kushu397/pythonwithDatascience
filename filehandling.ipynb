### Dictionary Data to CSV File


```python
import csv

products = [
    {
        'Product Name': 'Laptop',
        'Product Price': '$10000',
        'Product Brand': 'HP'
    },
    {
        'Product Name': 'Laptop',
        'Product Price': '$30000',
        'Product Brand': 'Lenovo'
    },
    {
        'Product Name': 'Laptop',
        'Product Price': '$20000',
        'Product Brand': 'Mac'
    }
]

# Create File
with open('product.csv', 'w', newline='') as file:
    columns = ['Product Name', 'Product Price', 'Product Brand']
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(products)

# File Read
with open('product.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        for key, val in row.items():
            print(f'{key}:{val}')
        print('\n')
```

```python
import csv

# Header and initial data
header = ['Item_ID', 'Item_Name', 'Stock_Quantity', 'Minimum_Required']
records = [
    ['101', 'Keyboard', '15', '10'],
    ['102', 'Mouse', '4', '8'],
    ['103', 'Monitor', '2', '5'],
    ['104', 'USB Cable', '25', '15']
]

# Write data to inventory.csv
with open('inventory.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerows(records)

print("inventory.csv initialized successfully.")
```

```python
###Task 2 :Automated LOw stock scan
```

```python
import csv

low_stock_items = []

# Read and scan inventory.csv
with open('inventory.csv', mode='r') as file:
    reader = csv.reader(file)
    header = next(reader)  # Skip header row
    
    for row in reader:
        stock_quantity = int(row[2])
        minimum_required = int(row[3])
        
        # Compare integer values
        if stock_quantity < minimum_required:
            low_stock_items.append(row)
            print(f"Low Stock Found: {row[1]} (Stock: {stock_quantity}, Min: {minimum_required})")
```

```python
### Task 3 Generate reorder list
```

```python
import csv

# Read low-stock items from inventory
low_stock_items = []

with open('inventory.csv', mode='r') as file:
    reader = csv.reader(file)
    header = next(reader)
    
    for row in reader:
        if int(row[2]) < int(row[3]):
            low_stock_items.append(row)

# Write to reorder_list.csv
with open('reorder_list.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerows(low_stock_items)

print("Reorder list generated successfully.")
```
