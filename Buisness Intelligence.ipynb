```python
# Case 1: Executive Payroll Allocation
# The Chief Financial Officer (CFO) is reviewing the company's financial allocations for the 
# upcoming fiscal year. She wants to identify which job roles carry the heaviest financial weight and 
# where corporate capital is concentrated.
# • Task: Generate a report showing the total cash allocation, the typical payout, and the total 
# workforce size for every distinct position in the company. Organize the output so that the 
# positions costing the company the most money appear at the very top.
```

```python
import pandas as pd

df = pd.read_csv('cleaned_hr.csv')
df.columns
```

```python
payroll_report = (
    df.groupby('Position')
    .agg(
        Total_Cash_Allocation=('Salary', 'sum'),
        Typical_Payout=('Salary', 'mean'),
        Workforce_Size=('EmpID', 'count')
    )
    .reset_index()
    .sort_values(by=['Total_Cash_Allocation'], ascending=False,ignore_index=True)
    .rename(columns={
        'Position': 'Position',
        'Total_Cash_Allocation': 'Total Cash Allocation',
        'Typical_Payout': 'Typical Payout',
        'Workforce_Size': 'Workforce Size'
    })
)
payroll_report['Total Cash Allocation'] = payroll_report['Total Cash Allocation'].map('${:,.2f}'.format)
payroll_report['Typical Payout'] = payroll_report['Typical Payout'].map('${:,.2f}'.format)

payroll_report
```

```python
# Case 2: Workplace Attendance & Engagement Audit
# The Operations Team suspects that employee burnout and high absenteeism might be linked to 
# drops in motivation or strict performance standards.
# • Task: Extract a metric profile for each performance rating tier. For each tier, find the 
# typical number of absences, the standard survey score for employee commitment, and the 
# worst-case scenario for short-term tardiness (DaysLateLast30)
```

```python
attendance_audit = df.groupby('PerformanceScore').agg(
    Typical_Absences=('Absences', 'median'),
    Commitment_Score=('EngagementSurvey', 'mean'),
    Worst_Case_Tardiness=('DaysLateLast30', 'max')
).reset_index()

attedance_audit = attendance_audit.rename(columns={
    'PerformanceScore':'Performace Score',
    'Typical_Absences': 'Typical Absences',
    'Commitment_Score':'Commitment Score',
    'Worst_Case_Tardiness':'Worst Case Tardiness'
})
attendance_audit['Commitment_Score']=attendance_audit['Commitment_Score'].round(2)
attendance_audit
```

```python
# Case 3: Talent Acquisition Pipeline Evaluation
# The Recruiting Director is evaluating the historical success of various hiring channels to optimize 
# next quarter's advertising budget. He wants to see if certain channels pull in more talent or 
# negotiate higher baselines, broken down by demographic representation.
# • Task: Build a structured lookup table that breaks down the total volume of hired staff and 
# their typical compensation package based on how they were recruited and their gender. 
# Ensure the final result is outputted as a clean, flat table format
```

```python
# df.head()
```

```python
structured_lookup = df.groupby(['RecruitmentSource', 'Gender']).agg(
    Total_Hired=('EmpID', 'count'),
    Typical_compensation=('Salary', 'mean')
).reset_index()
structured_lookup = structured_lookup.rename(columns={
    'RecruitmentSource':'Recruitment Source',
    'Gender':'Gender',
    'Total_Hired':'Total Hired',
    'Typical_compensation':'Typical Compensation'
})

structured_lookup['Typical Compensation']=structured_lookup['Typical Compensation'].round(2)
structured_lookup
```

```python
# Case 4: Leadership & Team Health Assessment
# The VP of Human Resources wants to run an audit on the internal management structure to identify 
# which team leaders might be facing systemic team disengagement or operational bottlenecks.
# • Task: Create a leaderboard showing the collective absenteeism and the overall job 
# satisfaction landscape for the teams sitting under each specific supervisor. Sort the final 
# data to immediately highlight the supervisor whose team reports the lowest baseline 
# happiness
```

```python
import pandas as pd

leaderboard = df.groupby('ManagerID').agg(
    Total_absenteeism = ('Absences','sum'),
    EmpSatisfaction=('EmpSatisfaction','mean'),
    Total_Team=('EmpID','count')
).reset_index().sort_values(
    by=['Total_absenteeism', 'ManagerID', 'EmpSatisfaction'], 
    ascending=True, 
    ignore_index=True
)


leaderboard = leaderboard.rename(columns={
    'ManagerID': 'Manager ID',
    'Total_absenteeism':'Total Absenteeism',
    'EmpSatisfaction':'EmpSatisfaction',
    'Total_Team':'Total Team'
})

leaderboard['EmpSatisfaction'] = leaderboard['EmpSatisfaction'].round(2)
leaderboard
```
