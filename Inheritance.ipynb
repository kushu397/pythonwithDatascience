### Single Inheritance

```python
class Customer:
    def __init__(self, customer_name, customer_address, customer_phone, customer_email, customer_id):
        self.customer_name = customer_name
        self.customer_address = customer_address
        self.customer_phone = customer_phone
        self.customer_email = customer_email
        self.customer_id = customer_id

    def customerDetails(self):
        print(f"Customer_name: {self.customer_name}")
        print(f"Customer_address: {self.customer_address}")
        print(f"Customer_phone: {self.customer_phone}")
        print(f"Customer_email: {self.customer_email}")
        print(f"Customer_id: {self.customer_id}")


# Child class
class Order(Customer):
    def __init__(self,order_id,order_date,order_destination,order_status,customer_name,customer_address,customer_phone,customer_email,customer_id ):
        
        super().__init__(customer_name,customer_address,customer_phone,customer_email,customer_id)
        self.order_id = order_id
        self.order_date = order_date
        self.order_destination = order_destination
        self.order_status = order_status

    def orderDetails(self):
        print(f"Order_id: {self.order_id}")
        print(f"Order_date: {self.order_date}")
        print(f"Order_destination: {self.order_destination}")
        print(f"Order_status: {self.order_status}")


orders = []

o = Order(
    104,
    "2026-08-26",        # order_date
    "Kathmandu",         # order_destination
    "Completed",         # order_status
    "bob",           # customer_name
    "Jhapa",             # customer_address
    "12234584895",       # customer_phone
    "charles@gmail.com", # customer_email
    "cust11"             # customer_id
)

orders.append(o)

o.orderDetails()
print('\n')
o.customerDetails()


def findCustomerOrders(customer_name):
    for order in orders:
        if customer_name == order.customer_name:
            order.orderDetails()   # FIXED


findCustomerOrders('bob')
```

```python
class Customer:
    def __init__(self, customer_name, customer_address, customer_phone, customer_email, customer_id):
        self.customer_name = customer_name
        self.customer_address = customer_address
        self.customer_phone = customer_phone
        self.customer_email = customer_email
        self.customer_id = customer_id

    def customerDetails(self):
        print(f"Customer_name: {self.customer_name}")
        print(f"Customer_address: {self.customer_address}")
        print(f"Customer_phone: {self.customer_phone}")
        print(f"Customer_email: {self.customer_email}")
        print(f"Customer_id: {self.customer_id}")
```

```python
class Category:
    def __init__(self, category_id, category_name):
        self.category_id = category_id
        self.category_name = category_name

    def orderCategories(self):
        print(f"Category_id: {self.category_id}")
        print(f"Category_name: {self.category_name}")
```

```python
class Product:
    def __init__(self, product_id, product_name, model, year, list_price):
        self.product_id = product_id
        self.product_name = product_name
        self.model = model
        self.year = year
        self.list_price = list_price

    def productDetails(self):
        print(f"Product_id: {self.product_id}")
        print(f"Product_name: {self.product_name}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")
        print(f"List_price: {self.list_price}")
```

```python
class Stock:
    def __init__(self, stock_id, product_id, quantity):
        self.stock_id = stock_id
        self.product_id = product_id
        self.quantity = quantity

    def stockDetails(self):
        print(f"Stock_id: {self.stock_id}")
        print(f"Product_id: {self.product_id}")
        print(f"Quantity: {self.quantity}")
        
```

```python
class Brand:
    def __init__(self, brand_id, brand_name):
        self.brand_id = brand_id
        self.brand_name = brand_name

    def brandDetails(self):
        print(f"Brand_id: {self.brand_id}")
        print(f"Brand_name: {self.brand_name}")
```

```python
class Order(Customer, Category, Product, Stock, Brand):
    def __init__(
        self,
        order_id,
        order_date,
        order_destination,
        order_status,
        category_id,
        category_name,
        product_id,
        product_name,
        model,
        year,
        list_price,
        stock_id,
        quantity,
        brand_id,
        brand_name,
        customer_name,
        customer_address,
        customer_phone,
        customer_email,
        customer_id
    ):

        Customer.__init__(
            self,
            customer_name,
            customer_address,
            customer_phone,
            customer_email,
            customer_id
        )

        Category.__init__(
            self,
            category_id,
            category_name
        )

        Product.__init__(
            self,
            product_id,
            product_name,
            model,
            year,
            list_price
        )

        Stock.__init__(
            self,
            stock_id,
            product_id,
            quantity
        )

        Brand.__init__(
            self,
            brand_id,
            brand_name
        )

        self.order_id = order_id
        self.order_date = order_date
        self.order_destination = order_destination
        self.order_status = order_status

    def orderDetails(self):
        print(f"Order_id: {self.order_id}")
        print(f"Order_date: {self.order_date}")
        print(f"Order_destination: {self.order_destination}")
        print(f"Order_status: {self.order_status}")
```

```python
class Category:
    def __init__(self, category_id, category_name):
        self.category_id = category_id
        self.category_name = category_name

    def orderCategories(self):
        print(f"Category_id: {self.category_id}")
        print(f"Category_name: {self.category_name}")
```

```python
class Brand:
    def __init__(self, brand_id, brand_name):
        self.brand_id = brand_id
        self.brand_name = brand_name

    def brandDetails(self):
        print(f"Brand_id: {self.brand_id}")
        print(f"Brand_name: {self.brand_name}")
```

```python
class Product(Category, Brand):
    def __init__(
        self,
        product_id,
        product_name,
        product_desc,
        product_price,
        category_id,
        category_name,
        brand_id,
        brand_name
    ):

        Category.__init__(self, category_id, category_name)
        Brand.__init__(self, brand_id, brand_name)

        self.product_id = product_id
        self.product_name = product_name
        self.product_desc = product_desc
        self.product_price = product_price

    def product_details(self):
        print(f"Product_id: {self.product_id}")
        print(f"Product_name: {self.product_name}")
        print(f"Product_desc: {self.product_desc}")
        print(f"Product_price: {self.product_price}")


   
```

```python
"""
Inheritance: mechanism where one class (child)
acquires the properties and methods of parent class.
Child class uses parent class's code

syntax:

class Person:
    pass

class Child(Parent):
    pass

"""
```

```python
class person:
    def show(self):
        print('This is a person')

        
class student(person):
    def display(self):
        print('This is student')

class employee(person):
    def print(self):
        print('This is employee')
        
s = student()
s.show()
s.display()
s = employee()
s.print()
```
