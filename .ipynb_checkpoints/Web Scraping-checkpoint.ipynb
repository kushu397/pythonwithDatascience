```python
from selenium import webdriver #access web browser
from selenium.webdriver.common.by import By #access web site element/tags
```

```python
website = 'https://merolagani.com/LatestMarket.aspx'
```

```python
# access web browser
#driver = webdriver.Edge()
driver = webdriver.Chrome()
```

```python
driver.get(website)
```

# Access Table body

```python
# Find_element(By=finders,element_name)
```

```python
# 2. Access the table body and extract all rows
table_body = driver.find_element(By.TAG_NAME, "tbody")
rows = table_body.find_elements(By.TAG_NAME, "tr")
```

```python

# for row in rows:
#     columns = row.find_elements(By.TAG_NAME,'td')

#     for col in columns:
#         print(col.text)
```

### List to store data

```python
symbol_col = []
ltp_col =[]
change_col =[]
open_col =[]
high_col = []
low_col = []
qty_col =[]
pclose_col = []
diff_col = []
```

```python
for row in rows:
    columns = row.find_elements(By.TAG_NAME,'td')

    symbol_col.append(columns[0].text)
    ltp_col.append(columns[1].text)
    change_col.append(columns[2].text)
    open_col.append(columns[3].text)
    high_col.append(columns[4].text)
    low_col.append(columns[5].text)
    qty_col.append(columns[6].text)
    pclose_col.append(columns[7].text)
    diff_col.append(columns[8].text)


# symbol_col
```

```python
data = {
    'Symbol': symbol_col,
    'LTP':ltp_col,
    '%Change':change_col,
    'Open':open_col,
    'High':high_col,
    'Low':low_col,
    'Qty':qty_col,
    'Pclose':pclose_col,
    'Diff': diff_col
    

}
```

```python
driver.quit()
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
# df
```
