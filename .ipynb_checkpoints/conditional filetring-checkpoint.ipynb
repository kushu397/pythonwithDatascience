```python
import pandas as pd
import numpy as np

# 1. Load the dataset (using correct relative path based on where your notebook is)
# If your notebook is inside the Dataset folder, use 'Employees.csv'
# If your notebook is outside the Dataset folder, use 'Dataset/Employees.csv'
df = pd.read_csv('Dataset/Employees.csv')

print("--- Original Data Shape ---")
print(df.shape)

# 2. DATA CLEANING
# ---------------------------------------------------------
# a. Handle Null Values in Salary (using exact column name: SALARY)
if 'SALARY' in df.columns:
    df['SALARY'] = df['SALARY'].fillna(df['SALARY'].mean())

# Drop rows if EMPLOYEE_ID is missing (using exact column name)
if 'EMPLOYEE_ID' in df.columns:
    df = df.dropna(subset=['EMPLOYEE_ID'])

# b. Handle Duplicate Values
print("\nDuplicate rows found:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after dropping duplicates:", df.shape)

# c. Clean Text Columns (Strip whitespace from FIRST_NAME or LAST_NAME)
for col in ['FIRST_NAME', 'LAST_NAME', 'EMAIL', 'JOB_ID']:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()


# 3. CHECK STATISTICAL DATA TYPES & SKEWNESS
# ---------------------------------------------------------
print("\n--- Skewness Check for Numeric Columns ---")
# Select numeric columns (like SALARY, COMMISSION_PCT, MANAGER_ID, DEPARTMENT_ID)
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    skew_value = df[col].skew()
    print(f"Skewness of {col}: {skew_value:.2f}")
    
    # If skewness is high, apply log transformation
    if skew_value > 1.0:
        df[col + '_log'] = np.log1p(df[col])
        print(f"   -> Applied log transformation to {col}")

print("\n--- Cleaned DataFrame Head ---")
df.head()
```
