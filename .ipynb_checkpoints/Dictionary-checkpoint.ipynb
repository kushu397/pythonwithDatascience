### Dictionary Looping

```python
'''
- Values()-> loops only value of dictionary
-keys()-> loops only keys of dictonary
-items()->loops both keys and values of dictionary
'''
```

```python
for key in product.keys():
    print(key)
```

```python
for key,val in product.items():
    print(f'{key}: {val}')
```

```python
employee = {
    "name": "Sita",
    "department": "IT",
    "salary": 45000,
    "experience": 2
}

for key, value in employee.items():
    print(key, ":", value)
```

```python
product_details ={

    'product_name':['Mobile','Charger','Watch'],
    'product brand':['Samsung','Ankar','NavyForce'],
    'product proce':['Rs.50000','Rs.5000','Rs.35000'],
    'quantity':[3,2,4]
}
product_details
```

```python
for key , val in product_details.items():
    print(f'{key}:{val}')
    for idx,val in enumerate(val):
        print(f'{idx+1}:{val}')
        print('\n')
```

```python
# Lab Question 2: Advanced Slicing & Analysis Given the list: numbers = [12, 45, 2, 89, 33, 
# 10, 55, 78, 90, 6]
# 1. Print a slice of the list containing only the middle four elements.
# 2. Use slicing to print the list in reverse order.
# 3. Extract every third element starting from the first one.
# 4. Find and print the length, maximum value, and minimum value of this list using 
# Python’s built-in functions
```

```python
numbers = [12,45,2,89,33,10,55,78,90,6]
#Middle foyr elements
print("Middle four elements:",numbers[3:9])
#Reverse order list
print("Reverse Order list:",numbers[::-1])
#Extract every third element
print("Extract every third element:", numbers[::3])
#Find length,maximum,minimum
print("Length:",len(numbers))
print("Maximum:",max(numbers))
print("Minimum:",min(numbers))
```

```python
# Lab Question 3: Student Database Update Create a dictionary named student with the 
# following keys: "name": "Alice", "age": 20, and "major": "Physics".
# 1. Add a new key-value pair: "gpa": 3.8.
# 2. Update the "major" to "Computer Science".
# 3. Delete the "age" key using the pop() method.
# 4. Loop through the dictionary and print only the keys.
# 5. Loop through the dictionary using .items() and print the details in the format: Field: 
# Value
```

```python
student = {
    "name": "Alice",
    "age": 20,
    "major": "Physics"
}

# 1. Add GPA
student["gpa"] = 3.8

# 2. Update major
student["major"] = "Computer Science"

# 3. Delete age using pop()
student.pop("age")

# 4. Loop through dictionary and print only keys
print("Keys:")
for key in student:
    print(key)

# 5. Loop using items() and print Field: Value
print("\nStudent Details:")
for key, value in student.items():
    print(f"Field: {key}, Value: {value}")
```

```python
# Create a dictionary called cafeteria_menu where the keys are categories (e.g., 
# "Snacks", "Drinks") and the values are lists of items.
# 2. Add a new item to the "Drinks" list using the .append() method inside the dictionary.
# 3. Use a for loop and the enumerate() function to print the list of "Snacks" with their 
# index numbers (starting from 1).
# 4. Access and print the second item from the "Snacks" list using double indexing (e.g., 
# dict[key][index])
```

```python
cafeteria_menu = {

'Snacks': ['Noodles','pizza','burger']
    'Drinks':['coke','sprite']
    
}
#add new item
cafeteria_menu['Drinks'].append('Sprite')
```

```python

cafeteria_menu = {
    'Snacks': ['Noodles', 'pizza', 'burger'],
    'Drinks': ['coke', 'sprite']
}

# Add new item
cafeteria_menu['Drinks'].append('Sprite')

# Display menu
for key, value in cafeteria_menu.items():
    print(key)
    print('_' * 10)

    for idx, val in enumerate(value):
        print(f'{idx + 1}: {val}')

    print('\n')
```

```python
# Lab Question 7: The Gradebook (Dictionary with Nested Lists) Create a dictionary 
# named gradebook: gradebook = {"Sam": [80, 85, 90], "Tara": [92, 88, 95]}
# 1. Add a new student named "Leo" with scores [70, 75, 80].
# 2. Update Sam's second score (index 1) from 85 to 88.
# 3. Delete the last score in Tara's list using the pop() method.
# 4. Use a for loop and .items() to print each student's name and their highest score 
# using the max() function
gradebook = {
    "Sam": [80, 85, 90],
    "Tara": [92, 88, 95]
}

# 1. Add a new student
gradebook["Leo"] = [70, 75, 80]

# 2. Update Sam's second score
gradebook["Sam"][1] = 88

# 3. Delete Tara's last score
gradebook["Tara"].pop()

# 4. Print each student's highest score
for key ,val in gradebook.items():
    print(f'{key}:{max(val)}')

```

### Student Gradebook

```python
gradebook = {
    "Sam": [80, 85, 90],
    "Tara": [92, 88, 95],
    "Leo": [70, 75, 80]
}
```

```python
gradebook = {
    "Sam": [80, 85, 90],
    "Tara": [92, 88, 95],
    "Leo": [70, 75, 80]
}

# 1. Add Riya
gradebook["Riya"] = [85, 90, 95]

# 2. Update Leo's third score
gradebook["Leo"][2] = 85

# 3. Delete Sam's first score
gradebook["Sam"].pop(0)

# 4. Print total marks
for name, scores in gradebook.items():
    print(name, "Total Marks:", sum(scores))
```

#### Bank Transactions

```python
transactions = {
    "Sam": [1000, 2000, 1500],
    "Tara": [3000, 2500, 4000]
}
transactions = {
    "Sam": [1000, 2000, 1500],
    "Tara": [3000, 2500, 4000]
}

# 1. Add Leo
transactions["Leo"] = [1200, 1800, 2000]

# 2. Update Sam's second transaction
transactions["Sam"][1] = 2500

# 3. Delete Tara's last transaction
transactions["Tara"].pop()

# 4. Print total transactions
for name, amounts in transactions.items():
    print(name, "Total:", sum(amounts))
```

### Movie Ratings

```python
movies = {
    "Avatar": [8, 9, 10],
    "Titanic": [9, 8, 9]
}
movies = {
    "Avatar": [8, 9, 10],
    "Titanic": [9, 8, 9]
}

# 1. Add Inception
movies["Inception"] = [9, 10, 9]

# 2. Update Avatar's second rating
movies["Avatar"][1] = 10

# 3. Delete Titanic's last rating
movies["Titanic"].pop()

# 4. Print highest rating
for movie, ratings in movies.items():
    print(movie, "Highest Rating:", max(ratings))
```
