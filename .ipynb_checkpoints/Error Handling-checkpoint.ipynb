```python
# 1. The "Bank" Class
# Create a class called Bank. This class will be the "blueprint" for every account. It must 
# have:
# • Attributes:
# o Name: To store the account holder's name.
# o Balance: To store how much money is in the account.
# o Account Number: A unique 16-digit number generated automatically using 
# the random module.
# • Actions (Methods):
# o Deposit: Add a specific amount of money to the balance.
# o Withdraw: Subtract money from the balance (only if the user has enough).
# o Show Details: Print the name, balance, and account number to the screen.
    
```

# Bank App

```python
#Import Modules nd Custom Errors
import random


# Custom exception errors to handle bad inputs cleanly
class DepositAmountError(Exception):
    pass


class WithdrawAmountError(Exception):
    pass


class AccountNotFoundError(Exception):
    pass
```

```python
#Define the Bank Account Class
class Bank:

    def __init__(self, name, balance, job, staff_id):
        self.name = name
        self.balance = balance
        self.job = job
        self.staff_id = staff_id

        # Generates a simple 10-digit random account number
        self.account_number = str(random.randint(1000000000, 9999999999))

    def deposit(self, amount):
        if amount > 100:
            self.balance += amount
            print(
                f"Rs.{amount} deposited successfully to A/C: {self.account_number}"
            )
        else:
            raise DepositAmountError("Deposit amount must be greater than Rs. 100.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(
                f"Rs.{amount} withdrawn successfully from A/C: {self.account_number}"
            )
        else:
            raise WithdrawAmountError(
                "Withdrawal amount exceeds current balance."
            )

    def show_details(self):
        print("\n--- USER ACCOUNT DETAILS ---")
        print(f"Name           : {self.name}")
        print(f"Account Number : {self.account_number}")
        print(f"Balance        : Rs.{self.balance}")
        print(f"Job Position   : {self.job}")
        print(f"Verified By    : Staff ID {self.staff_id}")
        print("----------------------------")
```

```python
#Setup Storage and Helper Search Function
# List to store all created customer accounts
all_accounts = []


# Helper function to find an account by its 10-digit number
def find_account(acc_num):
    for acc in all_accounts:
        if acc.account_number == acc_num:
            return acc
    raise AccountNotFoundError(f"Account number '{acc_num}' not found!")
```

```python
# Staff Login
print("====================================")
print("   BANK STAFF AUTHENTICATION SYSTEM ")
print("====================================")

logged_in_staff = None

while not logged_in_staff:
    staff_id = input("Enter Staff ID (STAFF101): ").strip()
    password = input("Enter Password (admin123): ").strip()

    if staff_id == "STAFF101" and password == "admin123":
        logged_in_staff = staff_id
        print(f"\n[LOGIN SUCCESS] Welcome Staff Member {logged_in_staff}!")
    else:
        print("[ERROR] Wrong Staff ID or Password! Try again.\n")
```

 # Main Banking Menu Loop

```python
while True:
    print("\n------------------------------------")
    print(f"  MAIN MENU (Staff Active: {logged_in_staff})")
    print("------------------------------------")
    print("1. Create Customer Account")
    print("2. Deposit Amount")
    print("3. Withdraw Amount")
    print("4. View Customer Details")
    print("5. Exit")

    choice = input("\nEnter choice (1-5): ").strip()

    # 1. Create Account
    if choice == "1":
        print("\n--- Create Account ---")
        name = input("Enter customer full name: ").strip()
        balance = float(input("Enter initial deposit: "))
        job = input("Enter customer job position: ").strip()

        if balance > 100:
            new_acc = Bank(name, balance, job, logged_in_staff)
            all_accounts.append(new_acc)

            print("\n-------------------------------------------")
            print("ACCOUNT CREATED & VERIFIED!")
            print(f"Account Holder : {new_acc.name}")
            print(f"Account Number : {new_acc.account_number}")
            print(f"Current Balance: Rs.{new_acc.balance}")
            print("-------------------------------------------")
        else:
            print("[ERROR] Initial deposit must be more than Rs. 100.")

    # 2. Deposit Amount
    elif choice == "2":
        print("\n--- Deposit Amount ---")
        acc_num = input("Enter account number: ").strip()
        try:
            acc = find_account(acc_num)
            amount = float(input("Enter deposit amount: "))
            acc.deposit(amount)
        except (AccountNotFoundError, DepositAmountError, ValueError) as e:
            print(f"[ERROR] {e}")

    # 3. Withdraw Amount
    elif choice == "3":
        print("\n--- Withdraw Amount ---")
        acc_num = input("Enter account number: ").strip()
        try:
            acc = find_account(acc_num)
            amount = float(input("Enter withdrawal amount: "))
            acc.withdraw(amount)
        except (AccountNotFoundError, WithdrawAmountError, ValueError) as e:
            print(f"[ERROR] {e}")

    # 4. View Account Details
    elif choice == "4":
        print("\n--- View Account Details ---")
        acc_num = input("Enter account number: ").strip()
        try:
            acc = find_account(acc_num)
            acc.show_details()
        except AccountNotFoundError as e:
            print(f"[ERROR] {e}")

    # 5. Exit
    elif choice == "5":
        print("\nExited system. Goodbye!")
        break

    else:
        print("[ERROR] Invalid choice! Select 1 to 5.")
        
```
