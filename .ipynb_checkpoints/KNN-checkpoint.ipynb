```python
df['G3'].sort_values(ascending=False).unique()
```

```python
def convertMarks(marks):
    if marks >= 15 and marks <= 20:
        return 'A'
    elif marks >= 11  and marks < 15:
        return 'B'
    elif marks >= 7 and marks < 11:
        return 'C'
    else:
        return 'Fail'
```

```python
df['Grade'] = df['G3'].apply(convertMarks)
```

```python
# Function for manual encoding
def manualEncoding(dataframe, col_name, *args):
    encoding_val ={}
    for idx, data in enumerate(args):
        encoding_val[data] = idx

    dataframe[f'{col_name}_id'] = dataframe[col_name].map(encoding_val)
    print(f'{col_name} data encoded.')
```

```python
# Fedu 
manualEncoding(df, 'Fedu','none','primary education(4th grade)','5th to 9th grade','secondary education','higher education')
```

```python
# Travel time
manualEncoding(df, 'traveltime','<15 min ','15 to 30min.','30min. to 1 hour','>1 hour')
```

```python
# study Time 
manualEncoding(df, 'studytime','2 hours','2 to 5 hours','5 to 10 hours','>10 hours')
```

```python
# free time and Go out
data = ['freetime','goout']
for col in data:
       manualEncoding(df,col,'very low','low','medium','high','very high')
```

```python
manualEncoding(df,'health','very bad','bad','good','very good','excellent')
```

```python
manualEncoding(df, 'Grade','Fail','C','B','A')
```

## Feature and Target

```python
features = ['medu_id', 'Fedu_id', 'traveltime_id',
       'studytime_id', 'failures', 'schoolsup_id', 'famsup_id','internet_id','freetime_id','goout_id','health_id', 'absences','G1', 'G2']
target = 'Grade_id'
```

```python
X = df[features]
Y = df[target]
```

# Train and test split

```python
X_train , X_test, Y_train , Y_test  = train_test_split(
X,Y, test_size = 0.2, random_state = 42, stratify=Y)
```

# Feature Scaling

```python
scaler = StandardScaler()
X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

# Elbow Method 

```python
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier

# 1. Impute missing values on your already-scaled data
imputer = KNNImputer(n_neighbors=5)
X_train_clean = imputer.fit_transform(X_train_scale)
X_test_clean = imputer.transform(X_test_scale)

# 2. Run your loop using the cleaned variables
k_range = range(1, 21)
error_rate = []
acc_score = []

for i in k_range:
  knn = KNeighborsClassifier(n_neighbors=i, n_jobs=1)
  knn.fit(X_train_clean, Y_train)
  knn_pred = knn.predict(X_test_clean)

  error_rate.append(np.mean(Y_test != knn_pred))
  acc_score.append(accuracy_score(Y_test, knn_pred))
```

```python
fig , axes = plt.subplots(1,2,figsize=(14,4))

ax1 = axes[0]
ax1.plot(k_range, error_rate, marker='o')
ax1.set_xlabel('K Range')
ax1.set_ylabel('Error Rate')
ax1.set_title('Error Rate - KNN Finding Optimal')

ax2 = axes[1]
ax2.plot(k_range, error_rate, marker='o')
ax2.set_xlabel('K Range')
ax2.set_ylabel('Accuracy Score')
ax2.set_title(' - KNN Finding Optimal')
```

# Model Implementation

```python
k = 5
model = KNeighborsClassifier(n_neighbors=k, n_jobs=1)
model.fit(X_train_scale, Y_train)
Y_pred = model.predict(X_test_scale)
```

# Metrics

```python
cr = classification_report(Y_test, Y_pred)
cm = confusion_matrix(Y_test, Y_pred)
```

# Grid Serach

```python
'''
-euclidean -> continuous data + dense data,
-manhattan -> ordinal data 
-euclidean + manhattan -> nominal data
'''
params={
'n_neighbors': list(range(1,21)),
'weights':['uniform','distance'],
'metrics':['eulidean','manhattan']   
}
```
