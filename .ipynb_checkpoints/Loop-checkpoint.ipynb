```python

'''
#password
#Uppercase->isupper()
#Lowercase->islower
#digit->isdigit()
#special character
#min>8/max>16
len(text)

'''




#password
correct_password = "Password123!"
attempts = 0

while attempts < 3:
    password = input("Enter your password: ")

        #special character
        #min>8/max>16
          #len(text)
    if not (8 <= len(password) <= 16):
        print(
            "Invalid password format! Password must be between 8 and 16 characters."
        )
        attempts += 1
        continue

    # Check for Uppercase, Lowercase, Digit, and Special Character
    has_upper = any(char.isupper() for char in password)
    has_lower = any(char.islower() for char in password)
    has_digit = any(char.isdigit() for char in password)
    has_special = any(char in "!\"#$%&'()*+,-./:;<=>?@[\\]^_`{|}~" for char in password)

    if not (has_upper and has_lower and has_digit and has_special):
        print(
            "Invalid password format! Must contain at least one uppercase, "
            "one lowercase, one digit, and one special character."
        )
        attempts += 1
        continue

    # Check if password matches
    if password == correct_password:
        print("Login successful!")
        break
        
    else:
        print("Incorrect password.")
        attempts += 1

if attempts == 3:
    print("Account locked!")
```

```python
'''
while
if/elif
for
break
continue

1.Add product
2. View product
3.Search product
4. Exit

'''



cart = []

while True:

  
    print("1. Add product")
    print("2. View products")
    print("3. Search product")
    print("4. Exit")

    choice = input("Enter your choice: ")

    
    if choice == "1":

        product = input("Enter product name: ")

        cart.append(product)

        print(product, "added to cart.")

  
    elif choice == "2":

        if len(cart) == 0:
            print("Your cart is empty.")

        else:
            print("\nYour products:")

            for product in cart:
                print("-", product)

 
    elif choice == "3":

        search = input("Enter product to search: ")

        found = False

        for product in cart:

            if product.lower() == search.lower():
                found = True
                break

        if found:
            print("Product found!")

        else:
            print("Product not found.")

   
    elif choice == "4":

        print("Thank you for shopping!")
        break

    
    else:

        print("Invalid choice. Please try again.")
        continue
```

```python
'''
While
if/elif
break
continue
1. Check Balance
2. Deposit
3. Withdraw
4. Exit

'''




balance = 10000

while True:

   
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        print("Your balance is:", balance)


    elif choice == "2":

        amount = float(input("Enter deposit amount: "))

        if amount <= 0:
            print("Invalid deposit amount.")
            continue

        balance += amount

        print("Deposit successful.")
        print("New balance:", balance)

    
    elif choice == "3":

        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid withdrawal amount.")
            continue

        if amount > balance:
            print("Insufficient balance.")
            continue

        balance -= amount

        print("Withdrawal successful.")
        print("Remaining balance:", balance)

    
    elif choice == "4":

        print("Thank you for using the ATM.")
        break


    else:

        print("Invalid choice.")
        continue
```

### E-Commerce Dynamic Shipping Engine


```python
cart_value = float(input("Enter cart value: "))

if cart_value > 150:
    print("Free shipping.")
else:
    print("Region: Domestic / International")
    region = input("Enter your region: ").capitalize()

    print("Membership: Yes / No")
    membership = input("Do you have any membership? ").capitalize()

    # Handle Membership check
    if membership == "Yes":
        print("Discount amount: $5")
        cart_value -= 5
        print(f"Cart Value After Discount: ${cart_value}")

        # Handle Region inside Membership = Yes
        if region == "Domestic":
            shipping_charge = 10
            cart_value += shipping_charge
            print(f"Shipping Charge: ${shipping_charge}")
            print(f"Cart Value after shipping charge: ${cart_value}")
        elif region == "International":
            shipping_charge = 25
            cart_value += shipping_charge
            print(f"Shipping Charge: ${shipping_charge}")
            print(f"Cart Value after shipping charge: ${cart_value}")
        else:
            print("Invalid Region. Please enter valid region.")

    elif membership == "No":
        # Handle Region inside Membership = No
        if region == "Domestic":
            shipping_charge = 10
            cart_value += shipping_charge
            print(f"Shipping Charge: ${shipping_charge}")
            print(f"Cart Value after shipping charge: ${cart_value}")
        elif region == "International":
            shipping_charge = 25
            cart_value += shipping_charge
            print(f"Shipping Charge: ${shipping_charge}")
            print(f"Cart Value after shipping charge: ${cart_value}")
        else:
            print("Invalid Region. Please enter valid region.")

    else:
        print("Invalid Membership Value")
```

### Automated Password Compliance validator

```python
'''
1. Password must be  min 8 character long and max 16 character long 
-> Special Charcater - Regular Expressions (Regex)
2. Password must containan at least 1uppercase, 1lowercase ,1 special characters, 1digit/number
3.Must not contain any space between.
'''



password = input("Enter your password: ")
pwd_length = len(password)

is_upper = False
is_lower = False
is_special = False
is_digit = False
is_space = False

special = " @#!$%&:;\"'"

if 8 <= pwd_length <= 16:
    for char in password:
        if char.isupper():  # check uppercase
            is_upper = True
        if char.islower():  # check lowercase
            is_lower = True
        if char.isdigit():  # check digit/number
            is_digit = True
        if char in special:  # check special character
            is_special = True
        if char == ' ':  # check space
            is_space = True

    # Validate that all conditions were met
    if is_upper and is_lower and is_digit and is_special and not is_space:
        print("Password is valid!")
    else:
        print("Password must contain upper, lower, digit, special char, and NO spaces.")

    if is_space:
        print('Password must not contain any space character.')
    if not is_upper:
        print('Password must contain at least 1 uppercase character.')
    if not is_lower:
        print('Password must contain at least 1 lowerrcase character.')
    if not is_special:
        print('Password must contain at least 1 special character.')
    if not is_digit:
        print('Password must contain at least 1 digit character.')

else:
    print("Password must be minimum 8 characters long and maximum 16 characters long.")

print(is_space)
print(is_upper)
print(is_lower)
print(is_special)
print(is_digit)
```

```python
command = ""
started = False

while True:
    command = input("> ").lower()

    if command == "start":
        if started:
            print("Car is already started")
        else:
            started = True
            print("Car started")

    elif command == "stop":
        if not started:
            print("Car is already stopped!")
        else:
            started = False
            print("Car stopped")

    elif command == "help":
        print("""
Start - to start the car
Stop - to stop the car
Quit - to quit
""")

    elif command == "quit":
        break

    else:
        print("Sorry, I don't understand that!")
```
