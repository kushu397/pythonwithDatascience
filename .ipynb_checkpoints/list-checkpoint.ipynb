```python
#lab Question 6: Matrix Modification (2D lists) Given this 3*3 matrix: matrix=[[1,2,3],[4,5,6],[7,8,9]]
#1. change the middle value(5)to 0 using its roe and column index.
#2. add a nes row[10 ,11,12]to the wnd of the matrix
#3. Access the last element of the second row (6) ana first element of the douth row(10) and print their sum
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# 1. Change 5 to 0
matrix[1][1] = 0

# 2. Add new row
matrix.append([10, 11, 12])

# 3. Add 6 and 10
result = matrix[1][2] + matrix[3][0]

print(matrix)
print(result)
```

```python
#lab question5: The "Price Is Right"(conditonal Logic) you are give a list of product prce : prices =[100,250,75,400,120,310].
#1.Write a program that dreate twio nes empty lists: cheap_items and expensive _items
# 2. loop through yhe prices list.if a prices is greater than 200,append it to expensives_items.otherwise append to its cheap_items.
#     3. print both list and calculates the total count of items in each using the len() function
```

```python
prices = [100, 250, 75, 400, 120, 310]

cheap_items = []
expensive_items = []

# for price in prices:
#     if price >= 200:
#         expensive_items.append(price)
#     else:
#         cheap_items.append(price)
for price in prices:
    expensive_items.append(price) if price >= 200 else cheap_items.append(price) 

print("Cheap items:", cheap_items)
print("Expensive items:", expensive_items)

print("Total cheap items:", len(cheap_items))
print("Total expensive items:", len(expensive_items))
```

```python
# data = [1,2,2,2,3,3,1,4,5,5,5]
# duplicate_data=[]
# unique_data =[]
```

```python

data = [1,2,2,2,3,3,1,4,5,5,5]
duplicate_data=[]
unique_data =[]
for item in data:
    if data.count(item) > 1:
        if item not in duplicate_data:
            duplicate_data.append(item)
    else:
        unique_data.append(item)

print("Duplicate data:", duplicate_data)
print("Unique data:", unique_data)
```

```python
for val in data:
    if val in unique_data:#[1,2,3]#False
        if val not in duplicate_data:#[2]#true
            duplicate_data.append(val)
    else:
        unique_data.append(val)
print(unique_data)
print(duplicate_data)
```
