## All Arguments(*args) and all keyword argument(**kwargs)

### *args

```python
def total_marks(*marks):#dynamically afaii value ma gayerw basxs
    total = sum(marks)
    return total

result = total_marks(80,7,90,)
print('Total:',result)
```

```python
def empdetails(*args):
   for val in args:
    print(val)
empdetails(1,'bob','harry')
```

```python
def fooddetails(*args):
   for val in args:
       print(val)
fooddetails('momo','pizza','mangoshake')
```

```python
def combo_set(*food):
    total = (food)
    return total
result = combo_set('pizaa','momo','noodles','hot spicychicken wings','legchickenpiece')
print('Total:',result)
    
```

### **Kwargs

```python
def student_info(**details):

    for key,val in details.items():
       print(f'{key}:{val}')

student_info(
    name ='Kushu',
    age='23',
    address='kathmandu',
    course ='maths'
)
```

```python
def emdetails(**kwargs):
  print(kwargs)
  for key, val in kwargs.items():
         print(f'{key}:{val}')
emdetails(id = 1,name ='bob',position ='guffadi')
```

```python
def test(*args, **kwargs):
    print(args)
    print(kwargs)

test(10, 20, 30, name="Ram", age=20)
```

## program related to *args and **kwargs

```python
def student_result(name, *marks, **details):

    print("\n===== STUDENT RESULT =====")

    print("Name:", name)

    print("\nMarks:")

    for i, mark in enumerate(marks, start=1):
        print(f"Subject {i}: {mark}")

    total = sum(marks)
    average = total / len(marks)

    print("\nTotal:", total)
    print("Average:", average)

    print("\nAdditional Details:")

    for key, value in details.items():
        print(f"{key}: {value}")


student_result(
    "Kushu",
    80,
    75,
    90,
    85,
    95,
    course="BSc CSIT",
    semester="8th",
    college="TU"
)
```

### Lambda Function

```python
#One line function / Arrow Function
# var_name = lambda arg1,arg2: logic.../print..
```

```python
def add(a,b):
    return f'Addition of {a} and {b} is: {a+b}'
add(3,4)
```

```python
addition = lambda a , b : f'Addition of {a} and {b} is:{a+b}'
add(3,5)
```

```python
odd_even = lambda val : f'{val}:Even'if val % 2 == 0 else f'{val}:ODD'
print(odd_even(7))
print(odd_even(8))
```

```python
def odd_even():
    for i  in range (1,21):
       if i % 2 == 0:
          print(f'{i}:Even')
    else:
        print(f'{i}:Odd')
# odd_even()
```

```python
for_odd_even = lambda : [f'{i}:Even' if i % 2 == 0 else f'{i}:Odd' for i in range(1,21)]
for_odd_even()
```

### Nested function

```python

def add(a,b):
    return f'Addition of{a} and {b} is {a+b}'
    
def sub(a,b):
    return f'subtraction of{a} and {b} is {a-b}'
def mul(a,b):
    return f'Multiplication of{a} and {b} is {a*b}'
def div(a,b):
    return f'Divide of{a} and {b} is {a/b}'


def calculator():
    x = int(input('Enter your first number:'))
    y = int(input('Enter your second number:'))

    choice = int(input('Enter your choice:'))
    if choice == 1:
          print(add(x,y))
    elif choice == 2:
          print(sub(x,y))
    elif choice == 3:
          print(mul(x,y))
    elif choice == 2:
          print(div(x,y))
    else:
          print('Invalid choice:')
calculator()
```

### Without .split()

```python
data = input('Enter values: ')

result = {}
index = 0

for char in data:
    if char == ',':
        continue

    if char not in result:
        result[char] = index
        index += 1

print(result)
```

### with .split()

```python
user_input = input('Enter values: ')

items = user_input.split(',')

result = {}

index = 0

for item in items:
    for char in item:
        if char not in result:
            result[char] = index
            index += 1

print(result)
```

## setedefault

```python
data = input('Enter values: ')

result = {}
index = 0

for char in data:
    if char != ',':
        if char not in result:
            result.setdefault(char, index)
            index += 1

print(result)
```

### enumerate

```python
data = input('Enter value:')

result = {}

for index, char in enumerate(data):
    if char != ',' and char not in result:
         result[char] = len(result)
print(result)
```
