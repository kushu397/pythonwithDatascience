# Linear Regression

```python
'''
- Straight line concept
-y -> mX + C
-y -> Predicted Value
-M -> Slope(B1)
-X-> Data Points
-C -> Intercept /Coffiencient (constant)(B0)

-Eculidean Distance



Metrics
---------
1.MAE (Mean Absolute Error)
-> Distance between each data  points and best fit line
-> Sum of all distance
->(X2-X1)2 +(Y2-Y1)2

2.MSE(Mean Squared Error)
-> Squared distance between each data points and best fit line
-> sum of all squared distance
->((x2-x1)2+ (y2-y1)2)

3.RMSE(Root Mean Squared Error)-Loss Function
->Root(MSE)
->Root((x2-x1)2)+(y2-y1)2)

4.R2 Score(Resuadile Score)-Cost Function -> Accuracy

Rule
-----------
- if RMSE Increase R2 Score Decreases
- if RMSE Decreases R2 Score Increases -> Correct


Linear Regression Formula
---------------------------
-> if qw have multiple features
mlr = B0 + B1.X1 + B2.X2 + ... + Bn.Xn + E

-> If we have single feature
slr = B0 + B1.X1 + E


suppose 4 features -> different 4 slopes(B1) ans singles Intercept(B0)



Types of Rgeression
-----------------
1.polynomial Rgression
2. Ridge Regression
3.Lasso Rgression
4.ELASTIC NET regression
'''
```
