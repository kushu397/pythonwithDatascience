```python
import sqlite3
```

```python
conn = sqlite3.connect('ecommerce.db')
```

```python
cursor = conn.cursor()
```

# Create Table

## Customr Table

```python
try:
    cursor.execute('''
     -- Customer Table
    CREATE TABLE Customers (
       customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
       customer_name TEXT NOT NULL,
       contact TEXT NOT NULL UNIQUE,
       email TEXT NOT NULL UNIQUE,
       customer_type TEXT NOT NULL
        CHECK (customer_type IN ('VIP', 'Regular', 'Non-Regular'))
    );
  ''')
    print('Customer table created successfully')
except sqlite3.OperationalError as e:
    print(e)
```

# Staff Table

```python
try: 
    cursor.execute(''' 
        Create Table Staffs( 
             staff_id integer primary key autoincrement, 
             staff_name text not null, 
             staff_phone text not null unique, 
             staff_email text not null unique, 
             staff_position text not null, 
             staff_type text not null, 
             constraint chk_staff_position check (staff_position in ('Senior','Junior','Associate')), 
             constraint chk_staff_type check (staff_type in ('Full-time','Part_time','Contract')) 
         
        ); 
         
    ''') 
    print('Staff table is created successfully') 
 
except sqlite3.OperationalError as e: 
    print(e)
```

# ORDER TABLE

```python
try:
    cursor.execute('''
      Create Table Orders(
           order_id integer primary key autoincrement,
           --yyyy-mm-dd
           order_date date not null,
           order_status not null,
           order_destination not null,
           customer_id integer,
           staff_id integer,
           constraint chk_order_status check(order_status in ('Completed','Pending','Rejected','Processing')),
           foreign key (customer_id) references customers (customer_id),
           foreign key (staff_id) references staffs (staff_id) 
      
      );
    
    ''')
    print('Order Table successfuly created.')
except sqlite3.OperationalError as e:
    print(e)
```
