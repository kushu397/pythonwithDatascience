# Seed:ufuncs

```python
import numpy as np

# Create array
B = np.random.rand(100_000_000)

# Time NumPy's function (runs once)
%time _ = np.sum(B)

# Time NumPy's method (runs once)
%time _ = B.sum()
```

```python
def mySum(G):
     s = 0
     for x in G:
          s+=x
     return s
```

```python
%timeit mySum(B)
```
