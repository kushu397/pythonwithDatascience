### Restuarant Order System

```python
print("Welcome to Restaurant Management System")

print('1. Breakfast')
print('2. Lunch')
print('3. Snacks')
print('4. Dinner')
print('5. Dessert')

choice = int(input('Enter your order choice (1-5): '))

match choice:
    case 1:
        print('You have ordered breakfast.')
        print('1. Tea')
        print('2. Egg')
        print('3. Bread')

        bf_choice = int(input('Enter your breakfast choice (1-3): '))

        match bf_choice:
            case 1:
                print('You have ordered Tea.')
            case 2:
                print('You have ordered Egg.')
            case 3:
                print('You have ordered Bread.')
            case _:
                print('Invalid Breakfast choice. Try again.')

    case 2:
        print('You have ordered Lunch.')

    case 3:
        print('You have ordered Snacks.')

    case 4:
        print('You have ordered Dinner.')

    case 5:
        print('You have ordered Dessert.')

    case _:
        print('Invalid order choice. Please choose between 1 and 5.')
```

```python
food = input("Enter food name: ")

food = food.strip().lower()

if food == "pizza":
    print("Pizza - Rs. 500")
else:
    if food == "burger":
        print("Burger - Rs. 300")
    else:
        if food == "momo":
            print("Momo - Rs. 200")
        else:
            if food == "chowmein":
                print("Chowmein - Rs. 250")
            else:
                print("Food not available")
```
