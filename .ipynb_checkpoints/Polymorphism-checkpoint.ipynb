# Abstraction

```python
'''
1.parent class must be Abstract class 
2.Atleast one function  in abstract class must be abstract function.
3.Child class implments parent class data.
'''
```

```python
# ABC -> Abstract base class
from abc import ABC, abstractmethod
```

```python
class Bank(ABC):
    def user(self):
        return f'Username: Artur'

    def connection(self):
        return 'Connected to server...'
        
```

```python
class Mobilebanking(Bank):
    def statement(self):
        return 'User Statement'

    def user(self):
        return f'Username: Artur'
```

```python
mb = Mobilebanking()
```

```python
mb.user()
```

```python
# 3. Masking Sensitive Data
# Create a Credential class that stores a private __api_key. Implement a method called 
# get_masked_key() that returns only the first 3 and last 3 characters of the key, with asterisks 
# in between (e.g., ABC*********XYZ).
# (API Key = “ABC123456789XYZ”
```

```python
class Credential:
    def __init__(self, api_key):
        self.__api_key = api_key

    def get_masked_key(self):
        return self.__api_key[:3] + "*" * (len(self.__api_key) - 6) + self.__api_key[-3:]



credential = Credential("ABC1233243890380923XYZ")

print(credential.get_masked_key())
```

```python
class rectangle:
    def area(self, x, y):
        return x * y

class triangle:
    def area(self, x, y):
        return (x * y) / 2

r = rectangle()
print(f"Area: {r.area(10, 5)}")  # Output: Area: 50

t = triangle()
print(f"Area: {t.area(10, 5)}")   # Output: Area: 25.0
```
