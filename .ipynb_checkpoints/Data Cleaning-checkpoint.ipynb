# Data Cleaning

```python
'''
Process of Data Cleaning
------------------------
1. Handle Null Values
    a. Fill Null Values
        - .fillna()
    -  If data is continuous 
        -Mean
        -median
        1. If mean and median value of a column is equal or almost equal -> fill mean value
        2. Skewness -> .skew()
            -Right Skewness/Positive Skewness
                -> data_skew >0.5 -> fill median value
            -Left Skewness / Negative Skewness
                data_skew < 0 -> fill median value
            -Normal Distribution / Bell Curve
                - 0 -0.5 -> fill mean value 
        
    - If data is discrete/categorical
        -Mode
    - If data is continuous /discrete/categorical
        -Custom Record
    b. Drop Null Values
        - .dropna()
2. Drop Duplicate Values
    - Find duplicate records
        -> .duplicated()

    - Drop duplicate records
        -> .drop_duplicates()
3. Remove unwanted data
    a.Remove white space
        -> .str.strip()
    b.values(regular expressions,text,digits)
    - .str.replace()
'''
```

```python
'''
Pattern of Regular Expressions
------------------------------------
\d ->Digits 
\s ->Whitespace
\w ->Word
+ -> One or More 
* -> Zero or More

sub(pattern,replace_value,data_variable) ->remove data
findall(pattern,data)-> find and extract data 
'''
```

```python
import re ## regular expression
```

```python
text ='    Hello   !!  contact us  : 54454324 abc@gmail.com    def133@gmail.com  2343545####   ' 
```

```python
remove_whitespace=re.sub(r'\s+',' ',text)
```

```python
remove_whitespace.strip()      ## this is left or right whitespace
```

```python
remove_val=re.sub(r'[!#]','',remove_whitespace)
remove_val.strip()
```

```python
remove_email=re.sub(r'\w+@\w+.\w+','',remove_whitespace)
remove_email.strip()
```

```python
## find email 

email =re.findall(r'\w+@\w+.\w+',text)
```

```python
import pandas as pd

# 1. Load the dataset from your CSV file
df = pd.read_csv('cleaned_hr.csv')

# 2. Consolidate First Name (prioritizing _y and filling missing with _x)
if 'First_Name_y' in df.columns and 'First_Name_x' in df.columns:
    df['First_Name'] = df['First_Name_y'].fillna(df['First_Name_x'])
elif 'First_Name_y' in df.columns:
    df['First_Name'] = df['First_Name_y']
elif 'First_Name_x' in df.columns:
    df['First_Name'] = df['First_Name_x']

# 3. Consolidate Last Name
if 'Last_Name_y' in df.columns and 'Last_Name_x' in df.columns:
    df['Last_Name'] = df['Last_Name_y'].fillna(df['Last_Name_x'])
elif 'Last_Name_y' in df.columns:
    df['Last_Name'] = df['Last_Name_y']
elif 'Last_Name_x' in df.columns:
    df['Last_Name'] = df['Last_Name_x']

# 4. Handle Middle Name
if 'Middle_Name_y' in df.columns:
    df['Middle_Name'] = df['Middle_Name_y']
elif 'Middle_Name_x' in df.columns:
    df['Middle_Name'] = df['Middle_Name_x']

# 5. Drop the old suffixed columns safely
columns_to_drop = [
    'First_Name_x', 'First_Name_y', 
    'Last_Name_x', 'Last_Name_y', 
    'Middle_Name_y', 'Middle_Name_x'
]
df = df.drop(columns=[col for col in columns_to_drop if col in df.columns], errors='ignore')

# 6. Save the improved dataframe back to your file
df.to_csv('cleaned_hr.csv', index=False)

# 7. Verify the remaining columns
print("Name columns successfully consolidated!")
print(df[['First_Name', 'Middle_Name', 'Last_Name']].head(10))
```
