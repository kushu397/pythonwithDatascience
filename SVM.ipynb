```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# model
from sklearn.svm import SVC
# Train Test
from sklearn.model_selection import train_test_split
#prepocessing
from sklearn.preprocessing import StandardScaler
# metrics
from sklearn.metrics import classification_report , confusion_matrix
```

```python
df = pd.read_csv('Dataset/Cardiovascular_disease.csv')
```

```python
df
```

```python
# Convert age from dyas to years
df['age'] = df['age']/365
```

```python
# Removing Outliers
#sns.boxplot(df['height']) # 150 - 185
#sns.bosplot(df['weight'])# 45 - 120
# sns.boxplot(df['ap_hi'])# 90-190
# sns.boxplot(df['ap_lo'])# 50 - 90
df = df[
 (df['height'].between(150, 185)) &
(df['weight'].between(45, 120))&
    (df['ap_hi'].between(90, 190))&
    (df['ap_lo'].between(50,90))

    ]
```

```python
df_sample = df.sample(n=6000, random_state=42)
df_sample
```

# Feature and Target

```python
features = ['age','gender','height','weight','ap_hi','ap_lo','cholesterol','gluc','smoke','alco','active']
target = 'cardio'
```

```python
X = df_sample[features]
Y= df_sample[target]
```

# Train test

```python
X_train, X_test, Y_train,Y_test = train_test_split(
  X,Y, test_size =0.2, random_state = 42,  stratify=Y
    
)
```

# Feature Scaling

```python
scaler = StandardScaler() # (Xi = mean)/sd
X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

# Model Implementation

```python
model = SVC()
```

```python
model.fit(X_train_scale, Y_train)
```

```python
Y_pred = model.predict(X_test_scale)
```

# Metrics

```python
cm = confusion_matrix(Y_test, Y_pred)
cr = classification_report(Y_test,Y_pred)
```
