```python
import random

# Create SubjectS
subjects = [
   "Shahrukh Khan",
   "Virat Kohli",
   "Nirmal Sitharaman",
    "A Mumbai Cat",
    "A Group of Monkeys",
    "Prime Ministed Modi",
    "Auto Rickshow Driver from Delhi"   
]
actions = [

    "Launches",
    "Cancles",
    "eats",
    "declares war on",
    "orders",
    "celebrates"  
]
places_or_things = [
    "Cat Red Fort",
    "in mumbai Local Train",
    "a plote of samosa",
    "inside parliament",
    "at Ganga Ghat",
    "during IPL Match",  
]

# Strat the headline generation loop

while True:
    subject = random.choice(subjects)
    action = random.choice(subjects)
    place_or_thing = random.choice(places_or_things)#conctinated the choice

    headline = f'BREAKING NEWS:{subject} {action} {place_or_thing}'
    print('\n' + headline)


    user_input = input('\n D you want another headline?(yes/no).strip().lower()')
    if user_input == 'no':
        break

#print goodbye message
print('\n Thanks for using the Fake News Headline ')
```
